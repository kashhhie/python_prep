```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
age=np.array([3,23,24,25,26,27,28,29,25,26,28,23,25,27,22,28,49])
print(age)
```

```python
sns.boxplot(y=age,color="pink")
plt.show()
```

```python
print(np.mean(age))
print(np.median(age))
```

```python
Q1=np.quantile(age,0.25)
print(Q1)
```

```python
Q2=np.quantile(age,0.50)
print(Q2)
```

```python
Q3=np.quantile(age,0.75)
print(Q3)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-(1.5*IQR)
print(lower_fence)
```

```python
upper_fence=Q3+(1.5*IQR)
print(upper_fence)
```

```python
l=[]
for i in age:
    if i<lower_fence or i>upper_fence:
        l.append(i)
print(i)
```
